```python
from utils import file
import matplotlib.pyplot as plt 
import segmentation_models as sm
import tensorflow as tf
import tensorflow_io as tfio
from PIL import Image

TRAINING_IMAGE_PATH = 'training/images'
TRAINING_MASK_PATH = 'training/masks'

TESTING_IMAGE_PATH = 'testing/images'
TESTING_MASK_PATH = 'testing/masks'

trainingImagePaths = file.getSortedFilePaths(TRAINING_IMAGE_PATH)
trainingMaskPaths = file.getSortedFilePaths(TRAINING_MASK_PATH)

testingImagePaths = file.getSortedFilePaths(TESTING_IMAGE_PATH)
testingMaskPaths = file.getSortedFilePaths(TESTING_MASK_PATH)


# x - is the image and y - is the mask
x_train = [];
y_train = [];
x_test  = [];
y_test  = [];

for i in range(len(trainingImagePaths)):
    x_train.append(Image.open(trainingImagePaths[i]))
    y_train.append(Image.open(trainingMaskPaths[i]))

for i in range(len(testingImagePaths)):
    x_test.append(Image.open(testingImagePaths[i]))
    y_test.append(Image.open(testingMaskPaths[i]))

sm.set_framework('tf.keras')
BACKBONE   = 'efficientnetb0'
n_classes  = 1 
activation = 'sigmoid'

preprocess_input  = sm.get_preprocessing(BACKBONE)
x_train = preprocess_input(x_train)
x_test = preprocess_input(x_test)

model = sm.Unet(BACKBONE, encoder_weights='imagenet')
model.compile('Adam', loss=sm.losses.bce_jaccard_loss, metrics=[sm.metrics.iou_score])
```

```python
# fit model
model.fit(
    x=x_train,
    y=y_train,
    batch_size=16,
    epochs=100,
    validation_data=(x_test, y_test),
)
```
